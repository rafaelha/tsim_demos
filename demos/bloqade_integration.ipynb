```python
from bloqade import squin
from bloqade.tsim import Circuit
import tsim
import bloqade.stim
```

```python
@squin.kernel
def main():
    q = squin.qalloc(6)
    squin.broadcast.reset(q)
    squin.h(q[0])
    squin.t(q[0])
    squin.h(q[0])
    for i in range(5):
        if i % 2 == 0:
            squin.cx(q[0], q[i + 1])
        else:
            squin.cx(q[i + 1], q[0])
    m = squin.broadcast.measure(q)
    squin.annotate.set_detector([m[0], m[1]], [0,0])
    squin.annotate.set_detector([m[0], m[2]], [0,0])
    squin.annotate.set_observable([m[0], m[1], m[5]], [0,0])
```

```python
circuit = Circuit(main)
```

```python
circuit.diagram(height=200)
```

```python
circuit.diagram("zx");
```

```python
sampler = circuit.compile_detector_sampler()
detectors, observables = sampler.sample(shots=1_000_000, separate_observables=True)
```
