```python
import tsim
import zxparam as zx

# Internal Tsim helpers, used here to show the compilation steps (not public API).
from tsim.core.graph import transform_error_basis
from tsim.core.parse import parse_stim_circuit
```

```python
c = tsim.Circuit.from_file("../circuits/msd_5.stim")
```

```python
sampler = c.compile_detector_sampler()
```

```python
c.diagram("zx-dets")
```

```python
# Reduced sampling graph, with the error parameters in the basis the sampler uses
g = c.get_sampling_graph(sample_detectors=True)
zx.full_reduce(g)
g, _ = transform_error_basis(g, parse_stim_circuit(c.stim_circuit).num_error_bits)

# The only non-Clifford component; its outputs are 4 detectors and the observable
g0 = next(k for k in g.connected_components() if zx.tcount(k))
outputs = sorted(g0.outputs(), key=g0.port)
g0.plug_basis(outputs[0], 0)  # fix the first output to 0
for v in outputs[1:]:
    g0.discard(v)  # marginalize the others
zx.draw(g0)
```

```python
zx.full_reduce(g0)
zx.draw(g0)
```

```python
result = zx.decompose(g0)  # stabilizer-rank decomposition, default strategy "cat5"
print(len(result.terms))
```

```python
print("common factor:", result.common.format())
for term in result.terms:
    print("+", term.format())
```
